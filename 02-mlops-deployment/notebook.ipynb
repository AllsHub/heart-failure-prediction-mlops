```python
# Import library
import os
import sys

from tfx.orchestration import metadata
from tfx.types import standard_artifacts
from ml_metadata.proto import metadata_store_pb2
from tfx.orchestration.experimental.interactive.interactive_context import InteractiveContext

# Definisi konstanta dan path pipeline
PIPELINE_NAME = 'aldomp7-pipeline'
OUTPUT_BASE = 'output'
PIPELINE_ROOT = os.path.join(OUTPUT_BASE, PIPELINE_NAME)
METADATA_PATH = os.path.join(PIPELINE_ROOT, 'metadata.sqlite')
```

```python
# Menjalankan pipeline TFX menggunakan Apache Beam Orchestrator
!python local_pipeline.py
```

```python
# Inisialisasi InteractiveContext untuk visualisasi metadata
metadata_connection_config = metadata.sqlite_metadata_connection_config(METADATA_PATH)

context = InteractiveContext(
    pipeline_root=PIPELINE_ROOT,
    metadata_connection_config=metadata_connection_config
)
```

```python
# Mengambil dan memvisualisasikan artefak ExampleStatistics terbaru
with metadata.Metadata(metadata_connection_config) as store:
    stats_artifacts = store.get_artifacts_by_type('ExampleStatistics')
    
    if stats_artifacts:
        stats_gen_output = stats_artifacts[-1]
        print("Artifact ID:", stats_gen_output.id)
        print("Artifact URI:", stats_gen_output.uri)
        
        # Menampilkan visualisasi dari eksekusi terakhir
        context.show(stats_gen_output)
    else:
        print("Artifact ExampleStatistics tidak ditemukan.")
```

```python
# Mengambil dan memvisualisasikan artefak Schema terbaru
with metadata.Metadata(metadata_connection_config) as store:
    schema_artifacts = store.get_artifacts_by_type('Schema')
    
    if schema_artifacts:
        schema_output = schema_artifacts[-1]
        print("Artifact ID:", schema_output.id)
        
        context.show(schema_output)
    else:
        print("Artifact Schema tidak ditemukan.")
```

```python
# Mengambil dan memvisualisasikan hasil evaluasi model (TFMA) terbaru
with metadata.Metadata(metadata_connection_config) as store:
    model_eval_artifacts = store.get_artifacts_by_type('ModelEvaluation')
    
    if model_eval_artifacts:
        model_eval_output = model_eval_artifacts[-1]
        print("Artifact ID:", model_eval_output.id)
        
        context.show(model_eval_output)
    else:
        print("Artifact ModelEvaluation tidak ditemukan.")
```
